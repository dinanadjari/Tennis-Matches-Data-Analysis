```python
import zipfile
import os
```

```python
def extract_all(zip_path, out_dir):
    os.makedirs(out_dir, exist_ok=True)

    with zipfile.ZipFile(zip_path) as outer:
        for inner_name in outer.namelist():
            if inner_name.endswith('.zip'):
                with outer.open(inner_name) as inner_file:
                    with zipfile.ZipFile(inner_file) as inner_zip:
                        inner_zip.extractall(out_dir)

extract_all('../Data/input/tennis_data.zip', '../Data/input')
```
